```python
import os
os.makedirs("../data/text_files",exist_ok=True)
```

```python
sample_texts={
    "../data/text_files/python_intro.txt":"""Python programming intro 
    python is high level , interpreted lang known for it simplicity and readability. created by Guido van Rossum and first released in 1901.

    key features:
    - easy to learn and use
    - extensive standard library 
    - cross-platform compatibility
    - strong community support

    """,
    "../data/text_files/machineLearning.txt": """ Machine Learning  is a subset of artificial intelligence that enables systems to learn and improve
    from experirence without being explicitly programmed. It focus on developing computer programs that can access and use it to learn for themselves
    Types of machine learning :
    1. Supervised Learning: Learning with labeled data
    2. Unsupervised Learning: Finding patterns unlabeled data
    3. Reinforcement Learning: Learning through rewards and penalties
    """

}
for filePath, content in sample_texts.items():
    with open(filePath, 'w', encoding="utf-8") as f:
        f.write(content)
print("sample text files created")
```

```python
from langchain_community.document_loaders import TextLoader
loader=TextLoader("../data/text_files/python_intro.txt", encoding="utf-8")
document=loader.load()
print(document)
```

```python
from langchain_community.document_loaders import DirectoryLoader

dir_loader=DirectoryLoader(
    "../data/text_files",
    glob="**/*.txt",
    loader_cls=TextLoader,
    loader_kwargs={'encoding':'utf-8'},
    show_progress=False
)

documents=dir_loader.load()
documents
```
